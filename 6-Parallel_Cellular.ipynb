```python

# Sandpile Cellular Automata-Based Load Balancing
# Simplified simulation for distributed systems

# Number of computing resources (cells)
num_servers = 6

# Initial workload of each server
server_loads = [30, 5, 25, 10, 20, 5]

# Processing capacity of each server
server_speeds = [1, 1.5, 1, 2, 1, 1.5]

# Algorithm parameters
max_iterations = 20
max_transfer = 5
imbalance_threshold = 2

# Calculate normalized load
def calculate_load(load, speed):
    return load / speed


# Display initial workload
print("Initial Server Loads:")
for i in range(num_servers):
    print(f"Server {i + 1}: {server_loads[i]}")

initial_total = sum(server_loads)

# Cellular automata load balancing
for iteration in range(max_iterations):

    # Store changes for simultaneous updating
    load_changes = [0] * num_servers

    # Each server interacts with its neighbors
    for i in range(num_servers):

        left = (i - 1) % num_servers
        right = (i + 1) % num_servers

        # Find the neighbor with the lowest normalized load
        neighbor = min(
            [left, right],
            key=lambda j: calculate_load(
                server_loads[j], server_speeds[j]
            )
        )

        current_load = calculate_load(
            server_loads[i], server_speeds[i]
        )

        neighbor_load = calculate_load(
            server_loads[neighbor], server_speeds[neighbor]
        )

        difference = current_load - neighbor_load

        # Redistribute workload if imbalance is detected
        if difference > imbalance_threshold:

            transfer = min(
                max_transfer,
                server_loads[i],
                max(1, int(difference * server_speeds[i] / 2))
            )

            load_changes[i] -= transfer
            load_changes[neighbor] += transfer

    # Apply all workload changes simultaneously
    for i in range(num_servers):
        server_loads[i] += load_changes[i]

    print(f"Iteration {iteration + 1}: {server_loads}")

    # Stop if no workload is transferred
    if sum(abs(change) for change in load_changes) == 0:
        break

# Display final results
final_total = sum(server_loads)

print("\nFinal Server Loads:")
for i in range(num_servers):
    print(f"Server {i + 1}: {server_loads[i]}")

print("\nPerformance Analysis:")
print("Initial Total Workload:", initial_total)
print("Final Total Workload:", final_total)

print("\nInitial Normalized Loads:")
for i in range(num_servers):
    print(
        f"Server {i + 1}: "
        f"{calculate_load([30, 5, 25, 10, 20, 5][i], server_speeds[i]):.2f}"
    )

print("\nFinal Normalized Loads:")
for i in range(num_servers):
    print(
        f"Server {i + 1}: "
        f"{calculate_load(server_loads[i], server_speeds[i]):.2f}"
    )

print("\nLoad balancing completed!")
```
