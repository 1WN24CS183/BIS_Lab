```python
import random

# Objective function
def function(x):
    return x ** 2


# Parameters
num_particles = 10
iterations = 50

w = 0.7
c1 = 1.5
c2 = 1.5

lower = -10
upper = 10

# Initialize particle positions and velocities
positions = [random.uniform(lower, upper) for _ in range(num_particles)]
velocities = [random.uniform(-1, 1) for _ in range(num_particles)]

# Initialize personal best
pbest_positions = positions.copy()
pbest_values = [function(x) for x in positions]

# Initialize global best
best_index = pbest_values.index(min(pbest_values))
gbest_position = pbest_positions[best_index]
gbest_value = pbest_values[best_index]


# Main PSO algorithm
for iteration in range(iterations):

    for i in range(num_particles):

        r1 = random.random()
        r2 = random.random()

        # Update velocity
        velocities[i] = (
            w * velocities[i]
            + c1 * r1 * (pbest_positions[i] - positions[i])
            + c2 * r2 * (gbest_position - positions[i])
        )

        # Update position
        positions[i] += velocities[i]

        # Keep position within limits
        positions[i] = max(lower, min(upper, positions[i]))

        # Calculate fitness
        value = function(positions[i])

        # Update personal best
        if value < pbest_values[i]:
            pbest_positions[i] = positions[i]
            pbest_values[i] = value

        # Update global best
        if value < gbest_value:
            gbest_position = positions[i]
            gbest_value = value


# Display result
print("Best Position:", gbest_position)
print("Minimum Function Value:", gbest_value)
```
